```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_regression
```

Data set for Linear Regression

```python
sample_size = 100
features_size = 1

x,y = make_regression(n_samples=sample_size, n_features=features_size, noise=12, random_state=42)

plt.scatter(x,y)
```

Values of the hyperparameters 

```python
batch_size = 1
epochs = 1000
alpha = 0.1
```

Gradient Descent

```python

weights = np.zeros(features_size, dtype=int).tolist()
bias = 0
loss = []

for i in range(epochs):
    y_pred = np.dot(x, weights) + bias
    gradient_descent = (2 / sample_size) * np.dot(x.T, y_pred - y)
    weights = weights - alpha * gradient_descent
    gradient_descent_bias = -2 * np.mean(y-y_pred)
    bias = bias - alpha * gradient_descent_bias
    loss.append(np.mean((y-y_pred)**2))

x_axis = np.arange(x.min(), x.max(), 0.1)
plt.scatter(x,y)
plt.plot(x_axis, weights[0]*x_axis + bias, color="red")
plt.title("Linear Regression")
plt.xlabel("x axis")
plt.ylabel("y axis")
plt.show()
```

```python
plt.scatter(y,y_pred)
x_axis = np.arange(y.min(), y.max(), 0.1)
plt.plot(x_axis, x_axis, color="red")
plt.title("Actual vs predicted")
plt.xlabel("Actual")
plt.ylabel("Predicted")
plt.show()
```

```python
plt.plot([i for i in range(epochs)], loss)
plt.title("Loss vs Epochs")
plt.xlabel("Epochs")
plt.ylabel("MSE (Loss)")
plt.show()
```
